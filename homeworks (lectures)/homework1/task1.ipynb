```python
import re

class EmptyTextError(Exception):
    pass

def analyse_text(text: str) -> str:
    if not isinstance(text, str) or text.strip().isdigit():
        raise TypeError("Текст має бути рядком")
    if text.strip() == "":
        raise EmptyTextError("Текст порожній")

    symbols = re.findall(r"[^\s]", text)

    return len(symbols)

text = input("Введіть ваш текст: ")

try:
    print(f"К-ть символів у тексті (без пробілів): {analyse_text(text)}")
except EmptyTextError:
    print("Введено пустий текст. Спробуйте ще раз")
except TypeError:
    print("Неправильний тип даних. Спробуйте ще раз")
```
