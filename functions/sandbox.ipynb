```python
sides = ["heads", "tails"]

heads_count = 0
tails_count = 0

choices = random.choices(sides, k=1000)
print(choices)

for choice in choices:
    if choice == "heads":
        heads_count += 1
    elif choice == "tails":
        tails_count += 1

print(f"К-ть разів, коли випав орел: {heads_count}")
print(f"К-ть разів, коли випала решка: {tails_count}")
```

```python
game_choices = ["камінь", "ножиці", "папір"]

other_choice = random.choice(game_choices)

print("""Вітаємо в грі 'Камінь-ножиці-папір'! Оберіть ваш хід черед запропонованих нижче:
1. камінь
2. ножиці
3. папір""")

user_choice = str(input("Введіть ваш хід:"))

if user_choice == "камінь" and other_choice == "ножиці":
    print("Ви виграли! :)")
elif user_choice == "камінь" and other_choice == "папір":
    print("Ви програли :(")
elif user_choice == "ножиці" and other_choice == "камінь":
    print("Ви програли :(")
elif user_choice == "ножиці" and other_choice == "папір":
    print("Ви виграли! :)")
elif user_choice == "папір" and other_choice == "камінь":
    print("Ви виграли! :)")
elif user_choice == "папір" and other_choice == "ножиці":
    print("Ви програли :(")
elif user_choice == other_choice:
    print("Нічия!")
else:
    print("Введіть корректний хід")

print(f"Хід вашого суперника: {other_choice}")
```
