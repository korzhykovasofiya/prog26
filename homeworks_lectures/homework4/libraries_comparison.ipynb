# Порівняльна таблиця бібліотек

| NLTK | spaCy |
|---|---|
| Навчально-дослідницька бібліотека для обробки природної мови (NLP), що містить набір інструментів, алгоритмів і корпусів текстів | Промислова бібліотека NLP для швидкої та зручної обробки великих обсягів тексту в реальних проєктах |
| Токенізація, стемінг, лематизація (WordNet), POS-теги, синтаксичний аналіз, робота з корпусами, частотні розподіли | Токенізація, POS-теги, лематизація, розпізнавання іменованих сутностей, синтаксичний аналіз, вектори слів, готові конвеєри обробки |
| Англійська підтримується повністю. Для української немає готових мовних моделей, але токенізатор на регулярних виразах (`RegexpTokenizer`) коректно працює з кирилицею | Англійська: готові моделі (`en_core_web_sm`). Українська: є токенізатор (`spacy.blank("uk")`) та окрема модель `uk_core_news_sm` |
| Простий для базових задач (токенізація, частотність), але частина функцій потребує окремого завантаження даних | Єдиний об'єкт `nlp` і зрозумілий API; повні моделі треба завантажувати окремо, для токенізації вони не потрібні |
| Легша в установленні, повільніша на великих текстах | Швидша (ядро написане на Cython), але важча, моделі можуть займати десятки мегабайтів |
