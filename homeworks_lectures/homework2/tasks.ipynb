```python
import datetime
import math
import random


current_time = datetime.datetime.now()
formatted_time = current_time.strftime("%Y-%m-%d %H:%M:%S")
print(f"Поточна дата і час: {formatted_time}")

random_number = random.randint(1, 100)
print(f"Випадкове число: {random_number}")

sqrt_value = math.sqrt(random_number)
sin_value = math.sin(math.radians(random_number))

print(f"Квадратний корінь з числа {random_number} дорівнює: {sqrt_value:.2f}")
print(f"Синус {random_number} градусів дорівнює: {sin_value:.2f}")
```
